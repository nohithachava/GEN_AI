```python

# ============================================================
# WEEK 6 - GENERATIVE AI PROGRAMMING PRACTICAL
# Llama 3.2 Baseline Evaluation - Release v0.1
# ============================================================
import os
import sys
import time
import json
import csv
import subprocess
import requests
from datetime import datetime
OLLAMA_URL = "http://localhost:11434/api/generate"
MODEL_NAME = "llama3.2"

print("=" * 60)
print("WEEK 6 - GENERATIVE AI PROGRAMMING PRACTICAL")
print("=" * 60)
subprocess.run("apt-get update -qq && apt-get install -y zstd", shell=True, check=True)

# STEP 1: INSTALL OLLAMA

print("\nSTEP 1: Installing Ollama...")

subprocess.run(
    "curl -fsSL https://ollama.com/install.sh | sh",
    shell=True,
    check=True
)

print("Ollama installed successfully!")

# STEP 2: START OLLAMA SERVER

print("\nSTEP 2: Starting Ollama server...")

ollama_process = subprocess.Popen(
    ["ollama", "serve"],
    stdout=subprocess.DEVNULL,
    stderr=subprocess.DEVNULL
)

OLLAMA_URL = "http://localhost:11434/api/generate"
MODEL_NAME = "llama3.2"

time.sleep(5)

# STEP 3: DOWNLOAD LLAMA 3.2

print("\nSTEP 3: Downloading Llama 3.2...")
print("This may take several minutes.")

subprocess.run(
    ["ollama", "pull", MODEL_NAME],
    check=True
)

print("Llama 3.2 downloaded successfully!")


# STEP 4: CREATE GOLDEN QUESTION SET

print("\nSTEP 4: Creating Golden Question Set...")

golden_questions = [
    {
        "id": "Q001",
        "question": "What is tokenization in large language models?",
        "expected_answer": "Tokenization is the process of splitting input text into smaller units called tokens.",
        "keywords": ["splitting", "text", "tokens"],
        "source": "https://huggingface.co/docs/transformers/en/tokenizer_summary"
    },
    {
        "id": "Q002",
        "question": "What is a token?",
        "expected_answer": "A token is a unit of text such as a word, subword, character or punctuation mark processed by a language model.",
        "keywords": ["unit", "text", "word", "subword"],
        "source": "https://huggingface.co/docs/transformers/en/tokenizer_summary"
    },
    {
        "id": "Q003",
        "question": "What is Byte Pair Encoding?",
        "expected_answer": "Byte Pair Encoding is a subword tokenization algorithm that repeatedly merges frequent adjacent symbol pairs.",
        "keywords": ["subword", "merges", "frequent", "pairs"],
        "source": "https://huggingface.co/docs/transformers/en/tokenizer_summary"
    },
    {
        "id": "Q004",
        "question": "What is WordPiece tokenization?",
        "expected_answer": "WordPiece is a subword tokenization algorithm that builds a vocabulary of frequently useful word pieces.",
        "keywords": ["subword", "vocabulary", "word", "pieces"],
        "source": "https://huggingface.co/docs/transformers/en/tokenizer_summary"
    },
    {
        "id": "Q005",
        "question": "What is SentencePiece?",
        "expected_answer": "SentencePiece is a language-independent tokenizer that can train directly on raw text and supports BPE and Unigram models.",
        "keywords": ["language-independent", "raw text", "BPE", "Unigram"],
        "source": "https://github.com/google/sentencepiece"
    },
    {
        "id": "Q006",
        "question": "What is the difference between BPE and WordPiece?",
        "expected_answer": "Both are subword tokenization methods, but BPE merges frequent symbol pairs while WordPiece uses a vocabulary learning approach based on the usefulness of word pieces.",
        "keywords": ["subword", "BPE", "WordPiece", "vocabulary"],
        "source": "https://huggingface.co/docs/transformers/en/tokenizer_summary"
    },
    {
        "id": "Q007",
        "question": "Why can Telugu text produce different token counts from English text?",
        "expected_answer": "Token counts depend on the tokenizer vocabulary, training data and how the language is represented in that vocabulary.",
        "keywords": ["tokenizer", "vocabulary", "language", "token counts"],
        "source": "https://huggingface.co/docs/transformers/en/tokenizer_summary"
    },
    {
        "id": "Q008",
        "question": "What is a token ratio?",
        "expected_answer": "Token ratio is the number of tokens divided by the number of words in the input text.",
        "keywords": ["tokens", "words", "divided"],
        "source": "https://huggingface.co/docs/transformers/en/tokenizer_summary"
    },
    {
        "id": "Q009",
        "question": "What are input tokens?",
        "expected_answer": "Input tokens are the tokens sent to a language model, including the prompt and other supplied context.",
        "keywords": ["input", "tokens", "prompt", "context"],
        "source": "https://ai.google.dev/gemini-api/docs/tokens"
    },
    {
        "id": "Q010",
        "question": "What are output tokens?",
        "expected_answer": "Output tokens are the tokens generated by a language model in its response.",
        "keywords": ["output", "tokens", "generated", "response"],
        "source": "https://ai.google.dev/gemini-api/docs/tokens"
    },
    {
        "id": "Q011",
        "question": "How is LLM API cost calculated?",
        "expected_answer": "API cost is calculated by multiplying input tokens by the input rate and output tokens by the output rate, then adding the two costs.",
        "keywords": ["input tokens", "output tokens", "input rate", "output rate"],
        "source": "https://ai.google.dev/gemini-api/docs/pricing"
    },
    {
        "id": "Q012",
        "question": "What is a context window?",
        "expected_answer": "A context window is the maximum amount of tokens a model can process in a single interaction, subject to the model's limits.",
        "keywords": ["context", "maximum", "tokens", "process"],
        "source": "https://huggingface.co/docs/transformers/en/model_doc/gpt2"
    },
    {
        "id": "Q013",
        "question": "What is a rate limit in an LLM API?",
        "expected_answer": "A rate limit restricts API usage over a period of time, such as requests per minute or tokens per minute.",
        "keywords": ["requests", "minute", "tokens", "limit"],
        "source": "https://ai.google.dev/gemini-api/docs/rate-limits"
    },
    {
        "id": "Q014",
        "question": "Why is monitoring token usage important?",
        "expected_answer": "Monitoring token usage helps estimate API costs, manage budgets and identify excessive usage.",
        "keywords": ["token usage", "costs", "budgets", "monitoring"],
        "source": "https://ai.google.dev/gemini-api/docs/tokens"
    },
    {
        "id": "Q015",
        "question": "What is autoregressive generation?",
        "expected_answer": "Autoregressive generation is the process of generating one token at a time using previously available tokens as context.",
        "keywords": ["one token", "previously", "context", "generating"],
        "source": "https://huggingface.co/docs/transformers/en/generation_strategies"
    },
    {
        "id": "Q016",
        "question": "What is next-token prediction?",
        "expected_answer": "Next-token prediction is the task of estimating the probability of possible next tokens based on the preceding sequence.",
        "keywords": ["probability", "next tokens", "preceding sequence"],
        "source": "https://huggingface.co/docs/transformers/en/generation_strategies"
    },
    {
        "id": "Q017",
        "question": "What is a probability distribution in language generation?",
        "expected_answer": "A probability distribution assigns probabilities to possible next tokens, with the probabilities summing to one.",
        "keywords": ["probabilities", "next tokens", "summing", "one"],
        "source": "https://huggingface.co/docs/transformers/en/generation_strategies"
    },
    {
        "id": "Q018",
        "question": "What is greedy decoding?",
        "expected_answer": "Greedy decoding selects the next token with the highest probability at each generation step.",
        "keywords": ["highest", "probability", "next token"],
        "source": "https://huggingface.co/docs/transformers/en/generation_strategies"
    },
    {
        "id": "Q019",
        "question": "What is sampling in text generation?",
        "expected_answer": "Sampling selects the next token according to the probability distribution rather than always selecting the highest probability token.",
        "keywords": ["sampling", "probability", "distribution", "token"],
        "source": "https://huggingface.co/docs/transformers/en/generation_strategies"
    },
    {
        "id": "Q020",
        "question": "What does temperature do during text generation?",
        "expected_answer": "Temperature controls the sharpness of the next-token probability distribution and influences the randomness of token selection.",
        "keywords": ["temperature", "probability", "distribution", "randomness"],
        "source": "https://huggingface.co/docs/transformers/en/main_classes/text_generation"
    },
    {
        "id": "Q021",
        "question": "What is top-k sampling?",
        "expected_answer": "Top-k sampling restricts token selection to the k tokens with the highest probabilities.",
        "keywords": ["top-k", "highest", "probabilities", "tokens"],
        "source": "https://huggingface.co/docs/transformers/en/main_classes/text_generation"
    },
    {
        "id": "Q022",
        "question": "What is top-p sampling?",
        "expected_answer": "Top-p sampling selects from the smallest set of highest-probability tokens whose cumulative probability reaches the specified threshold.",
        "keywords": ["top-p", "cumulative", "probability", "threshold"],
        "source": "https://huggingface.co/docs/transformers/en/main_classes/text_generation"
    },
    {
        "id": "Q023",
        "question": "What is an embedding in a language model?",
        "expected_answer": "An embedding is a numerical vector representation of a token or other input item.",
        "keywords": ["numerical", "vector", "representation", "token"],
        "source": "https://huggingface.co/docs/transformers/en/embeddings"
    },
    {
        "id": "Q024",
        "question": "What is a token ID?",
        "expected_answer": "A token ID is an integer that identifies a token in a tokenizer vocabulary.",
        "keywords": ["integer", "identifies", "token", "vocabulary"],
        "source": "https://huggingface.co/docs/transformers/en/tokenizer_summary"
    },
    {
        "id": "Q025",
        "question": "What is the difference between a token ID and an embedding?",
        "expected_answer": "A token ID is an integer identifier, while an embedding is a learned numerical vector representation associated with that token.",
        "keywords": ["integer", "identifier", "embedding", "vector"],
        "source": "https://huggingface.co/docs/transformers/en/embeddings"
    },
    {
        "id": "Q026",
        "question": "What is positional encoding?",
        "expected_answer": "Positional encoding provides information about the position or order of tokens in a sequence.",
        "keywords": ["position", "order", "tokens", "sequence"],
        "source": "https://arxiv.org/abs/1706.03762"
    },
    {
        "id": "Q027",
        "question": "Why does a Transformer need positional information?",
        "expected_answer": "Positional information helps a Transformer distinguish token order because the same words in different orders can have different meanings.",
        "keywords": ["positional", "token order", "different", "meanings"],
        "source": "https://arxiv.org/abs/1706.03762"
    },
    {
        "id": "Q028",
        "question": "What are sinusoidal positional encodings?",
        "expected_answer": "Sinusoidal positional encodings use sine and cosine functions at different frequencies to represent token positions.",
        "keywords": ["sine", "cosine", "frequencies", "positions"],
        "source": "https://arxiv.org/abs/1706.03762"
    },
    {
        "id": "Q029",
        "question": "What are Rotary Positional Embeddings?",
        "expected_answer": "Rotary Positional Embeddings, or RoPE, incorporate positional information by rotating token representations in the attention mechanism.",
        "keywords": ["RoPE", "positional", "rotating", "attention"],
        "source": "https://arxiv.org/abs/2104.09864"
    },
    {
        "id": "Q030",
        "question": "What is the role of the Transformer in an autoregressive language model?",
        "expected_answer": "The Transformer processes token representations using attention and neural network layers to produce information used for next-token prediction.",
        "keywords": ["Transformer", "attention", "token", "prediction"],
        "source": "https://arxiv.org/abs/1706.03762"
    }
]

with open("golden_set.json", "w", encoding="utf-8") as file:
    json.dump(golden_questions, file, indent=4, ensure_ascii=False)

print("Golden question set created!")
print("Total Questions:", len(golden_questions))

# STEP 5: FUNCTION TO GET LLM ANSWER

def get_llm_answer(question):

    payload = {
        "model": MODEL_NAME,
        "prompt": question,
        "stream": False,
        "options": {
            "temperature": 0
        }
    }

    response = requests.post(
        OLLAMA_URL,
        json=payload,
        timeout=300
    )

    response.raise_for_status()

    return response.json().get("response", "").strip()


# STEP 6: KEYWORD-BASED SCORING

def calculate_score(answer, expected_keywords):

    answer = answer.lower()

    matched_keywords = []
    missing_keywords = []

    for keyword in expected_keywords:

        if keyword.lower() in answer:
            matched_keywords.append(keyword)
        else:
            missing_keywords.append(keyword)

    total_keywords = len(expected_keywords)

    score = (
        len(matched_keywords) / total_keywords
    ) * 100 if total_keywords else 0

    return round(score, 2), matched_keywords, missing_keywords


# STEP 7: RUN BASELINE EVALUATION

results = []

total_score = 0
successful_questions = 0

print("\n" + "=" * 60)
print("BASELINE RUN STARTED")
print("=" * 60)

for index, item in enumerate(golden_questions, start=1):

    question_id = item["id"]
    question = item["question"]
    expected_answer = item["expected_answer"]
    keywords = item["keywords"]
    source = item["source"]

    print("\n" + "-" * 60)
    print(f"Question {index}: {question_id}")
    print("Question:", question)

    try:

        answer = get_llm_answer(question)

        score, matched, missing = calculate_score(
            answer,
            keywords
        )

        status = "PASS" if score >= 60 else "REVIEW"

        total_score += score
        successful_questions += 1

        print("\nLLM Answer:")
        print(answer)

        print("\nExpected Answer:")
        print(expected_answer)

        print("\nMatched Keywords:", matched)
        print("Missing Keywords:", missing)

        print("Score:", score, "%")
        print("Status:", status)

        results.append({
            "Question_ID": question_id,
            "Question": question,
            "Expected_Answer": expected_answer,
            "LLM_Answer": answer,
            "Matched_Keywords": ", ".join(matched),
            "Missing_Keywords": ", ".join(missing),
            "Score": score,
            "Status": status,
            "Source": source,
            "Model": MODEL_NAME,
            "Release": "v0.1"
        })

    except Exception as error:

        print("Error:", error)

        results.append({
            "Question_ID": question_id,
            "Question": question,
            "Expected_Answer": expected_answer,
            "LLM_Answer": "",
            "Matched_Keywords": "",
            "Missing_Keywords": ", ".join(keywords),
            "Score": 0,
            "Status": "ERROR",
            "Source": source,
            "Model": MODEL_NAME,
            "Release": "v0.1"
        })


# STEP 8: SAVE RESULTS TO CSV

OUTPUT_FILE = "baseline_results.csv"

fieldnames = [
    "Question_ID",
    "Question",
    "Expected_Answer",
    "LLM_Answer",
    "Matched_Keywords",
    "Missing_Keywords",
    "Score",
    "Status",
    "Source",
    "Model",
    "Release"
]

with open(
    OUTPUT_FILE,
    "w",
    newline="",
    encoding="utf-8-sig"
) as file:

    writer = csv.DictWriter(
        file,
        fieldnames=fieldnames
    )

    writer.writeheader()
    writer.writerows(results)


# STEP 9: BASELINE SUMMARY

if successful_questions > 0:
    average_score = total_score / successful_questions
else:
    average_score = 0

passed = sum(
    1 for result in results
    if result["Status"] == "PASS"
)

review = sum(
    1 for result in results
    if result["Status"] == "REVIEW"
)

errors = sum(
    1 for result in results
    if result["Status"] == "ERROR"
)

print("\n" + "=" * 60)
print("BASELINE EVALUATION SUMMARY")
print("=" * 60)

print("Release Version: v0.1")
print("Model:", MODEL_NAME)
print("Total Questions:", len(golden_questions))
print("Successful Responses:", successful_questions)
print("Passed Questions:", passed)
print("Questions for Review:", review)
print("Failed Requests:", errors)

print(
    "Average Keyword Coverage:",
    round(average_score, 2), "%"
)

print("Evaluation Date:", datetime.now())

# STEP 10: COMPLETION

print("\nResults saved to:", OUTPUT_FILE)
print("Golden set saved to: golden_set.json")

print("\nBASELINE RUN COMPLETED!")
print("Release v0.1 evaluation finished.")

# Download generated files in Colab

from google.colab import files

files.download("baseline_results.csv")
files.download("golden_set.json")
```
