```python

from collections import Counter

# 1. Input words and frequencies
words = {
    "low": 5,
    "lower": 2,
    "newest": 3,
    "wildest": 6
}

print("ORIGINAL WORDS:")
for word, freq in words.items():
    print(word, ":", freq)

# 2. Split words into characters
data = {tuple(word): freq for word, freq in words.items()}

print("\nINITIAL TOKENIZATION:")
for tokens, freq in data.items():
    print(" ".join(tokens), ":", freq)

# 3. Perform BPE merging
merge_count = 5

for iteration in range(merge_count):

    token_freq = Counter()
    pair_freq = Counter()

    # Calculate token and adjacent pair frequencies
    for tokens, freq in data.items():

        for token in tokens:
            token_freq[token] += freq

        for i in range(len(tokens) - 1):
            pair = (tokens[i], tokens[i + 1])
            pair_freq[pair] += freq

    # Calculate normalized pair scores
    scores = {}

    for pair, freq in pair_freq.items():
        first, second = pair

        score = freq / (token_freq[first] * token_freq[second])
        scores[pair] = score

    if not scores:
        break

    print(f"\nITERATION {iteration + 1}")
    print("\nPair Frequencies and Scores:")

    for pair, score in scores.items():
        print(pair, "Frequency:", pair_freq[pair],
              "Score:", round(score, 6))

    # Find the pair with the highest score
    best_pair = max(scores, key=scores.get)

    print("\nSelected Pair:", best_pair)
    print("Highest Score:", round(scores[best_pair], 6))

    # Merge the selected pair in all words
    new_data = {}

    for tokens, freq in data.items():

        merged = []
        i = 0

        while i < len(tokens):

            if (i < len(tokens) - 1 and
                (tokens[i], tokens[i + 1]) == best_pair):

                merged.append(tokens[i] + tokens[i + 1])
                i += 2

            else:
                merged.append(tokens[i])
                i += 1

        new_data[tuple(merged)] = freq

    data = new_data

    print("\nAfter Merging:")
    for tokens, freq in data.items():
        print(" ".join(tokens), ":", freq)

# 4. Display final tokenization
print("\n========== FINAL TOKENIZATION ==========")

for tokens, freq in data.items():
    print("Word:", "".join(tokens))
    print("Tokens:", list(tokens))
    print("Frequency:", freq)
    print()
```
