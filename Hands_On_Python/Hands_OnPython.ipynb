```python
import torch
from transformers import AutoTokenizer, AutoModelForCausalLM
# 1. Load pretrained GPT-2 tokenizer and causal language model
model_name = "gpt2"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForCausalLM.from_pretrained(model_name)
model.eval()
# 2. Encode input prompt into token IDs
text = "Artificial Intelligence is"
inputs = tokenizer(text, return_tensors="pt")
# 3. Perform one forward pass without gradient tracking
with torch.no_grad():
    outputs = model(**inputs)
# 4. Extract logits for the very last token position
logits = outputs.logits[:, -1, :]
# 5. Convert raw logits into probabilities via Softmax
probabilities = torch.softmax(logits, dim=-1)
# 6. Retrieve the top 5 most probable next tokens
top_probs, top_ids = torch.topk(probabilities, k=5)
print("Input:", text)
print("\nTop 5 Next Token Predictions:")
for prob, token_id in zip(top_probs[0], top_ids[0]):
    token = tokenizer.decode([token_id])
    print(f"{repr(token)} : {round(prob.item() * 100, 2)}%")
```
